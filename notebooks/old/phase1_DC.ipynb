```python
# Phase 1 – Data Collection
# Cell 2: Load & Initial Inspection of Macrohistory (JST R6)

import pandas as pd
import numpy as np
from pathlib import Path

# Paths (from Cell 1)
PROJECT = Path(r"D:\univercity\omid_project\economics_project")
MACRO_RAW = PROJECT / "data" / "raw" / "Macrohistory"
OUT_DATA  = PROJECT / "outputs" / "phase1" / "data"
OUT_TAB   = PROJECT / "outputs" / "phase1" / "tables"

print("=" * 60)
print("Phase 1 – Data Collection | Cell 2")
print("Loading Macrohistory (JST R6)...")
print("=" * 60)

# ---------------------- Load JST Dataset ----------------------
jst_path = MACRO_RAW / "JSTdatasetR6.xlsx"
df_jst = pd.read_excel(jst_path, sheet_name=0)

print(f"\nFile loaded: {jst_path.name}")
print(f"Shape      : {df_jst.shape[0]} rows × {df_jst.shape[1]} columns")
print(f"Memory     : {df_jst.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

# ---------------------- Basic Info ----------------------
print("\n--- Columns ---")
print(df_jst.columns.tolist())

print("\n--- First 5 rows ---")
print(df_jst.head())

print("\n--- Data Types ---")
print(df_jst.dtypes)

print("\n--- Year range ---")
print(f"Min year: {df_jst['year'].min()}")
print(f"Max year: {df_jst['year'].max()}")

print("\n--- Countries ---")
countries = sorted(df_jst['country'].unique())
print(f"Number of countries: {len(countries)}")
print(countries)

print("\n--- Missing values (top 15) ---")
missing = df_jst.isnull().mean().sort_values(ascending=False).head(15)
print(missing)

# ---------------------- Save raw copy + summary ----------------------
df_jst.to_pickle(OUT_DATA / "jst_raw.pkl")
df_jst.to_csv(OUT_DATA / "jst_raw.csv", index=False)

summary = pd.DataFrame({
    "Metric": ["Rows", "Columns", "Countries", "Year Min", "Year Max", "Memory (MB)"],
    "Value": [
        df_jst.shape[0],
        df_jst.shape[1],
        len(countries),
        df_jst['year'].min(),
        df_jst['year'].max(),
        round(df_jst.memory_usage(deep=True).sum() / 1024**2, 2)
    ]
})
summary.to_csv(OUT_TAB / "jst_summary.csv", index=False)

print(f"\nSaved:")
print(f"  → {OUT_DATA / 'jst_raw.pkl'}")
print(f"  → {OUT_DATA / 'jst_raw.csv'}")
print(f"  → {OUT_TAB / 'jst_summary.csv'}")
print("\nCell 2 completed successfully.")
```

```python
# Phase 1 – Data Collection
# Cell 3 (Corrected): Inspect IMF WEO files & Load the main dataset

import pandas as pd
import numpy as np
from pathlib import Path
from datetime import datetime

# Paths
PROJECT = Path(r"D:\univercity\omid_project\economics_project")
IMF_RAW = PROJECT / "data" / "raw" / "imf"          # <-- corrected
OUT_DATA = PROJECT / "outputs" / "phase1" / "data"
OUT_TAB  = PROJECT / "outputs" / "phase1" / "tables"

print("=" * 60)
print("Phase 1 – Data Collection | Cell 3 (Corrected)")
print("Inspecting IMF WEO files...")
print("=" * 60)

# ---------------------- List all IMF files ----------------------
print("\n--- Available files in imf ---")
if IMF_RAW.exists():
    for item in sorted(IMF_RAW.rglob("*")):
        if item.is_file():
            size_mb = item.stat().st_size / 1024**2
            print(f"{size_mb:8.2f} MB  |  {item.relative_to(IMF_RAW)}")
else:
    print("WARNING: imf folder not found!")

print("\n" + "-" * 60)

# ---------------------- Try to load the most useful file ----------------------
candidates = [
    IMF_RAW / "april 2026" / "New folder" / "WEOApr2026all.xlsx",
    IMF_RAW / "april 2026" / "dataset_2026-05-17T00_02_51.030039030Z_DEFAULT_INTEGRATION_IMF.RES_WEO_9.0.0.csv",
    IMF_RAW / "april 2025" / "all" / "weoapr2025all_By Countries.xls",
]

loaded = False
for path in candidates:
    if path.exists():
        print(f"\nTrying to load: {path.name}")
        print(f"Full path: {path}")
        
        try:
            if path.suffix.lower() in [".xlsx", ".xls"]:
                df_imf = pd.read_excel(path)
            else:
                # Large CSV – read carefully
                df_imf = pd.read_csv(path, low_memory=False)
            
            print(f"Successfully loaded!")
            print(f"Shape: {df_imf.shape[0]} rows × {df_imf.shape[1]} columns")
            
            print(f"\nColumns (first 25):")
            print(df_imf.columns.tolist()[:25])
            
            print(f"\nFirst 3 rows:")
            print(df_imf.head(3))
            
            # Save
            df_imf.to_pickle(OUT_DATA / "imf_weo_raw.pkl")
            print(f"\nSaved → {OUT_DATA / 'imf_weo_raw.pkl'}")
            
            loaded = True
            break
            
        except Exception as e:
            print(f"Failed to load {path.name}: {e}")
            continue

if not loaded:
    print("\nNo suitable IMF WEO file could be loaded.")

print("\nCell 3 completed.")
```

```python
# Phase 1 – Data Collection
# Cell 4: IMF Inventory + Final Phase 1 Summary

import pandas as pd
import numpy as np
from pathlib import Path
from datetime import datetime

PROJECT = Path(r"D:\univercity\omid_project\economics_project")
OUT_DATA = PROJECT / "outputs" / "phase1" / "data"
OUT_TAB  = PROJECT / "outputs" / "phase1" / "tables"

print("=" * 60)
print("Phase 1 – Data Collection | Cell 4")
print("Creating inventory and final summary...")
print("=" * 60)

# ---------------------- Load saved data ----------------------
df_jst = pd.read_pickle(OUT_DATA / "jst_raw.pkl")
df_imf = pd.read_pickle(OUT_DATA / "imf_weo_raw.pkl")

# ---------------------- IMF Inventory ----------------------
print("\n--- IMF Countries ---")
imf_countries = sorted(df_imf["COUNTRY"].dropna().unique())
print(f"Total countries/groups: {len(imf_countries)}")

print("\n--- IMF Indicators (unique) ---")
imf_indicators = df_imf["INDICATOR"].dropna().unique()
print(f"Total unique indicators: {len(imf_indicators)}")

# Save indicator list
ind_df = pd.DataFrame({"INDICATOR": sorted(imf_indicators)})
ind_df.to_csv(OUT_TAB / "imf_indicators_list.csv", index=False)
print(f"Saved indicator list → {OUT_TAB / 'imf_indicators_list.csv'}")

# Save country list
country_df = pd.DataFrame({"COUNTRY": imf_countries})
country_df.to_csv(OUT_TAB / "imf_countries_list.csv", index=False)
print(f"Saved country list → {OUT_TAB / 'imf_countries_list.csv'}")

# ---------------------- Match with JST 18 countries ----------------------
jst_countries = sorted(df_jst["country"].unique())
print(f"\nJST countries ({len(jst_countries)}):")
print(jst_countries)

# Simple name matching (can be refined later)
print("\n--- Checking presence of JST countries in IMF ---")
for c in jst_countries:
    found = any(c.lower() in str(ic).lower() or str(ic).lower() in c.lower() for ic in imf_countries)
    print(f"  {c:<15} → {'Found' if found else 'Not clearly found'}")

# ---------------------- Final Phase 1 Summary Table ----------------------
summary_data = {
    "Dataset": ["Macrohistory (JST R6)", "IMF WEO (April 2026)"],
    "Rows": [df_jst.shape[0], df_imf.shape[0]],
    "Columns": [df_jst.shape[1], df_imf.shape[1]],
    "Time Coverage": ["1870–2020", "1980–2031 (approx)"],
    "Countries": [len(jst_countries), len(imf_countries)],
    "File Saved": ["jst_raw.pkl", "imf_weo_raw.pkl"]
}
summary = pd.DataFrame(summary_data)
summary.to_csv(OUT_TAB / "phase1_final_summary.csv", index=False)

print("\n" + "=" * 60)
print("PHASE 1 FINAL SUMMARY")
print("=" * 60)
print(summary.to_string(index=False))
print("\nAll Phase 1 outputs saved in:")
print(f"  Data   : {OUT_DATA}")
print(f"  Tables : {OUT_TAB}")
print("\nCell 4 completed successfully.")
```
